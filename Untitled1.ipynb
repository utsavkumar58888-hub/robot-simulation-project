```python
%pip install stable_baselines3 gymnasium[classic_control,mujoco]

import gymnasium as gym
from stable_baselines3 import PPO
from stable_baselines3.common.evaluation import evaluate_policy
import matplotlib.pyplot as plt
import numpy as np

# ─────────────────────────────────────────
# 1. CREATE THE ENVIRONMENT
# ─────────────────────────────────────────
# We are using a robot arm environment called "Reacher"
# Goal: robot arm has to reach a target point
env = gym.make("Reacher-v4")

print("✅ Environment created!")
print(f"   Observation space: {env.observation_space}")
print(f"   Action space:      {env.action_space}")

# ─────────────────────────────────────────
# 2. UNTRAINED ROBOT — see how it performs BEFORE training
# ─────────────────────────────────────────
print("\n🤖 Testing UNTRAINED robot...")

obs, _ = env.reset()
untrained_rewards = []

for step in range(200):
    action = env.action_space.sample()  # random actions (no training)
    obs, reward, terminated, truncated, info = env.step(action)
    untrained_rewards.append(reward)
    if terminated or truncated:
        obs, _ = env.reset()

untrained_avg = np.mean(untrained_rewards)
print(f"   Average reward (untrained): {untrained_avg:.4f}")

# ─────────────────────────────────────────
# 3. TRAIN THE ROBOT using PPO algorithm
# ─────────────────────────────────────────
print("\n🏋️ Training the robot... (this will take a minute)")

model = PPO(
    "MlpPolicy",
    env,
    verbose=1,          # shows training progress
    learning_rate=0.001,
    n_steps=2048,
)

model.learn(total_timesteps=50000)  # train for 50,000 steps
print("\n✅ Training complete!")

# ─────────────────────────────────────────
# 4. TRAINED ROBOT — see how it performs AFTER training
# ─────────────────────────────────────────
print("\n🤖 Testing TRAINED robot...")

obs, _ = env.reset()
trained_rewards = []

for step in range(200):
    action, _ = model.predict(obs)  # use learned policy
    obs, reward, terminated, truncated, info = env.step(action)
    trained_rewards.append(reward)
    if terminated or truncated:
        obs, _ = env.reset()

trained_avg = np.mean(trained_rewards)
print(f"   Average reward (trained): {trained_avg:.4f}")

# ─────────────────────────────────────────
# 5. EVALUATE PROPERLY
# ─────────────────────────────────────────
mean_reward, std_reward = evaluate_policy(model, env, n_eval_episodes=10)
print(f"\n📊 Evaluation over 10 episodes:")
print(f"   Mean reward : {mean_reward:.4f}")
print(f"   Std  reward : {std_reward:.4f}")

# ─────────────────────────────────────────
# 6. VISUALIZE — Before vs After Training
# ─────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("3D Simulation for Training Robots\nBefore vs After Training", fontsize=14, fontweight='bold')

# Plot 1 — Reward over time (untrained)
axes[0].plot(untrained_rewards, color='red', alpha=0.7)
axes[0].axhline(y=untrained_avg, color='darkred', linestyle='--', label=f'Avg: {untrained_avg:.4f}')
axes[0].set_title("❌ Untrained Robot (Random Actions)")
axes[0].set_xlabel("Steps")
axes[0].set_ylabel("Reward")
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Plot 2 — Reward over time (trained)
axes[1].plot(trained_rewards, color='green', alpha=0.7)
axes[1].axhline(y=trained_avg, color='darkgreen', linestyle='--', label=f'Avg: {trained_avg:.4f}')
axes[1].set_title("✅ Trained Robot (PPO Algorithm)")
axes[1].set_xlabel("Steps")
axes[1].set_ylabel("Reward")
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig("robot_training_results.png", dpi=150, bbox_inches='tight')
plt.show()
print("\n📈 Graph saved as robot_training_results.png")

# ─────────────────────────────────────────
# 7. SUMMARY REPORT
# ─────────────────────────────────────────
improvement = trained_avg - untrained_avg
print("\n" + "="*50)
print("         📋 PROJECT SUMMARY REPORT")
print("="*50)
print(f"  Environment   : Reacher-v4 (Robot Arm)")
print(f"  Algorithm     : PPO (Proximal Policy Optimization)")
print(f"  Training Steps: 50,000")
print(f"  Untrained Avg Reward : {untrained_avg:.4f}")
print(f"  Trained Avg Reward   : {trained_avg:.4f}")
print(f"  Improvement          : {improvement:.4f}")
print(f"  Final Eval Reward    : {mean_reward:.4f} ± {std_reward:.4f}")
print("="*50)
print("\n✅ Project complete! You can show this output + graph in your presentation.")
```
